```python
import numpy as np
import matplotlib.pyplot as plt


def descente_gradient_SGD_par_lot(
    X0, y0, param, eta, max_iter=100, threshold=1e-3, lot=30
):
    for i in range(max_iter):
        rng = np.random.default_rng()
        indices = rng.choice(len(X0[:, 1]), size=lot, replace=False)
        X = X0[:, 1][indices]
        y = y0[indices]
        X = np.c_[np.ones((len(X), 1)), X.reshape(len(X), 1)]
        m = lot
        derive = (1 / m) * X.T @ (X @ param - y)
        print("Derive", derive)
        new_param = param - eta * derive
        if all(element < threshold for element in np.abs(derive)):
            print("Tous les paramétres ont convergé.")
            return new_param
        else:
            param = new_param
    print("Les paramétres n'ont pas convergé")
```

```python
np.random.seed(42)  # Pour la reproductibilité

# Nombre d'exemples
n_samples = 100
X = 2 * np.random.rand(n_samples, 1)
y = 4 + 3 * X + np.random.randn(n_samples, 1)

# Ajout d'une colonne de 1 pour le biais (theta0)
X_biais = np.c_[np.ones((m, 1)), X.reshape(m, 1)]  # forme (m, 2)
```

```python
# Visualisation des données
plt.scatter(X, y, alpha=0.7)
plt.title("Données synthétiques")
plt.xlabel("X")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

```python
param = descente_gradient_SGD_par_lot(
    X_biais,
    y,
    param=np.array([1, 5]).reshape(2, 1),
    eta=0.001,
    max_iter=100000,
    lot=70,
)
print(param)
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(X, y, label="Données (avec bruit)", alpha=0.6)
# Tracé de la droite y = θ0 + θ1*x
x_tracé = np.array([0, 10])
y_tracé = param[0] + param[1] * x_tracé
plt.plot(x_tracé, y_tracé, color="red", linewidth=2, label="Régression linéaire")
plt.title("Descente de gradient pour la régression linéaire")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```
