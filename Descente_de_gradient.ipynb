```python
import numpy as np
import matplotlib.pyplot as plt


def descente_gradient(X, y, param, eta, max_iter=100, threshold=1e-3):
    for i in range(max_iter):
        m = np.shape(X)[0]
        derive = (1 / m) * X.T @ (X @ param - y)
        print("Derive", derive)
        new_param = param - eta * derive
        if all(element < threshold for element in np.abs(derive)):
            print("Tous les paramétres ont convergé.")
            return new_param
        else:
            param = new_param
    print("Les paramétres n'ont pas convergé")
```

```python
np.random.seed(42)  # Pour la reproductibilité

# Nombre d'exemples
m = 100

# Génération de x uniformément dans [0, 10]
X = 10 * np.random.rand(m)

# Vraie relation linéaire : y = 3 * x + 4 + bruit
vrai_theta0 = 4
vrai_theta1 = 3
bruit = np.random.randn(m)  # bruit gaussien standard
y = vrai_theta0 + vrai_theta1 * X + bruit

# Ajout d'une colonne de 1 pour le biais (theta0)
X_biais = np.c_[np.ones((m, 1)), X.reshape(m, 1)]  # forme (m, 2)
np.shape(y)
```

```python
plt.scatter(X, y, color="blue", label="Données")
plt.xlabel("X")
plt.ylabel("y")
plt.title("Données générées")
plt.legend()
plt.show()
```

```python
param = descente_gradient(
    X_biais, y, param=np.array([1, 5]), eta=0.001, max_iter=100000
)
print(param)
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(X, y, label="Données (avec bruit)", alpha=0.6)
# Tracé de la droite y = θ0 + θ1*x
x_tracé = np.array([0, 10])
y_tracé = param[0] + param[1] * x_tracé
plt.plot(x_tracé, y_tracé, color="red", linewidth=2, label="Régression linéaire")
plt.title("Descente de gradient pour la régression linéaire")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```
